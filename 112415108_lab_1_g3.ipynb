# Working with CSV

Let's import a file with fuel economy data

```
import pandas as pd

mpg_df = pd.read_csv('mpg.csv')
mpg = mpg_df.to_dict('records')
cylinders = set(d['cyl'] for d in mpg)
MpgByCyl = []
for c in cylinders:
    summpg = 0
    typecount = 0
    for d in mpg:
        if d['cyl'] == c:
            summpg += float(d['cty'])
            typecount += 1

    if typecount > 0:
        avg_mpg = summpg / typecount
        MpgByCyl.append((int(c), avg_mpg))

MpgByCyl.sort(key=lambda x: x[0])
print("Cylinders and Avg City MPG:")
for cyl, avg_mpg in MpgByCyl:
    print(f"Cylinder {cyl}: {avg_mpg}")
print(type(mpg))
```
